#### Question 1.1: Write down the regression model you would try first.

 

  <h4 style="color:blue">(Write your answer here)</h4>


 

We first fit a logistic regression model with `dec` as the dependent variable (1 = sending a like, 0 = not sending a like).

The predictors are the participant's gender and age, together with their ratings of their partner's attractiveness (`attr`), sincerity (`sinc`), intelligence (`intel`), and fun (`fun`). We include interactions between gender and each numerical predictor to test whether these associations differ by gender.

The model specification is:

`logit[P(dec = 1)] = β₀ + β₁gender + β₂age + β₃attr + β₄sinc + β₅intel + β₆fun + β₇(gender × age) + β₈(gender × attr) + β₉(gender × sinc) + β₁₀(gender × intel) + β₁₁(gender × fun)`

All main effects are included alongside the interaction terms.

---

#### Question 1.2: In this model, which variables are numerical / nominal 

<h4 style="color:blue">(Write your answer here)</h4>

The numerical predictors are `age`, `attr`, `sinc`, `intel`, and `fun`. Age is measured in years, and the four trait ratings are treated as numerical scores.

The nominal predictor is `gender`, coded as 0 = female and 1 = male, with female as the reference group.

The dependent variable, `dec`, is binary: 1 = sending a like and 0 = not sending a like.

The interaction terms multiply gender by each numerical predictor, allowing the slopes to differ between genders.

```python
#----- set up the variables: 
design_int = MS([
    "gender", "age", "attr", "sinc", "intel", "fun",
    ("gender", "age"),
    ("gender", "attr"),
    ("gender", "sinc"),
    ("gender", "intel"),
    ("gender", "fun")
])

# Model specification and estimation
X_int = design_int.fit_transform(dating_sample_0)
y_int = dating_sample_0["dec"]

# Get results
glm_int = sm.GLM(
    y_int, X_int, family=sm.families.Binomial()
).fit()

# Result summary
glm_int_summary = summarize(glm_int)

# Significant and nonsignificant terms
glm_int_summary_sig = glm_int_summary[
    glm_int.pvalues < 0.1
]
glm_int_summary_insig = glm_int_summary[
    glm_int.pvalues >= 0.1
]

# Report results
print("significant table:\n", glm_int_summary_sig)
print("\nnonsignificant table:\n", glm_int_summary_insig)
```

### Question 2.4: 

In addition to the partner ratings and the trait-gap variables, what other explanatory variables would you include? Specify your regression model and justify your choices.
 

<h4 style="color:blue">(Write your answer here)</h4>

On top of the partner ratings and the gap variables, we add the participant’s gender and age to the model.

We include gender because it seems to matter for receiving likes. In our sample, women get a “yes” from their partner about 46% of the time, compared to about 39% for men. Gender also came up in Task 1 and in our gap chart in 2.2, where men had a bigger attractiveness gap than women, so it makes sense to control for it here too.

We include age because it could affect how a partner sees someone, and it’s a basic background variable that we already have in the cleaned data. Even if it doesn’t turn out to be significant, it’s reasonable to control for it.

The partner ratings  have to stay in the model because the gap variables are calculated from them. If a partner gives someone a low rating, that person’s gap gets bigger, and they’re also less likely to get a like. So without the partner ratings, the gap would mostly be picking up the effect of a low rating instead of overconfidence. With the partner ratings held constant, the gap coefficients tell us whether rating yourself higher is linked to getting a like, for people who were seen the same way by their partner.

We don’t add the self-ratings as separate variables. Since self-rating = gap + partner rating, including all three would cause perfect multicollinearity, and the model wouldn’t be able to separate their effects.

We also leave out the participant’s ratings of their partner. Those describe how the participant saw the partner, not how the partner saw the participant, so they don’t fit the question we’re asking about receiving a like.

### Question 2.7 — Independent Model Extension and Reflection

Please propose **one meaningful extension** to the regression model used in Question 2.5. There is no single correct extension.


Do not select an extension simply because it produces a statistically significant result.

**a) Motivation**

Describe your proposed extension. What question does it help you investigate, and why is it relevant? Before estimating the model, explain what result you expect and why.

In 2.5 we assumed the gap has the same effect for men and women. We want to check whether that’s true by adding an interaction between gender and each gap variable. The question is whether being overconfident hurts men and women differently when it comes to getting a like.

This is relevant because in 2.2 men had a bigger attractiveness gap than women, and in Task 1 we found that men and women don’t weigh traits the same way when they decide to send a like. For the same reason we expect the attractiveness gap to hurt women more, since their partners are men, who care more about attractiveness. We don’t expect much difference for sincerity or fun.

**b) Implementation**

Estimate the extended model. Clearly report:

- the regression;
- the modification you made;
- the relevant coefficient estimates and p-values; and
- any additional output needed to evaluate your extension.

We kept the same model as 2.5 and added four interaction terms: gender × attr_gap, gender × sinc_gap, gender × intel_gap and gender × fun_gap. With female = 0 as the reference group, the main gap coefficients show the effect for women, and each interaction shows how different the effect is for men.

```python
# write your code here

#======== Extension: do the gap effects differ by gender?
from scipy import stats
from sklearn.metrics import roc_auc_score

gap_cols = ["attr_gap", "sinc_gap", "intel_gap", "fun_gap"]

#--- same predictors as 2.5, plus gender x gap interactions
design_ext = MS(num_cols + [("gender", g) for g in gap_cols])

X_ext = design_ext.fit_transform(dating_sample_gap)
y_ext = dating_sample_gap["dec_o"]

glm_ext = sm.GLM(y_ext, X_ext, family=sm.families.Binomial()).fit()

#--- full regression output
print(glm_ext.summary())

#--- interaction terms
glm_ext_summary = summarize(glm_ext)
print("\nInteraction terms:\n", glm_ext_summary.loc[["gender:" + g for g in gap_cols]])

#--- gap effect for women vs men
slopes = pd.DataFrame({
    "Female": [glm_ext.params[g] for g in gap_cols],
    "Male": [glm_ext.params[g] + glm_ext.params["gender:" + g] for g in gap_cols]
}, index=gap_cols)
print("\nGap coefficients by gender:\n", slopes.round(4))

#--- compare with the 2.5 model
lr_stat = 2 * (glm_ext.llf - glm_overc.llf)
lr_p = stats.chi2.sf(lr_stat, df=4)

compare = pd.DataFrame({
    "Log-likelihood": [glm_overc.llf, glm_ext.llf],
    "AIC": [glm_overc.aic, glm_ext.aic],
    "AUC": [roc_auc_score(y_overc, glm_overc.predict()),
            roc_auc_score(y_ext, glm_ext.predict())]
}, index=["2.5 model", "Extended model"])
print("\nModel comparison:\n", compare.round(3))
print(f"\nLikelihood ratio test: statistic = {lr_stat:.2f}, df = 4, p-value = {lr_p:.4f}")
```

The interaction with gender is significant (−0.126, p = 0.004), which puts the effect for men at about −0.156. This means the intelligence result in 2.5 was mostly coming from men. The 2.5 model mixed men and women together, so it looked like an effect for everyone.

The attractiveness gap changed too. In 2.5 it wasn’t significant as p = 0.372, but now the interaction with gender is significant as p = 0.002. The effect is a bit negative for women −0.033 and a bit positive for men 0.080. Since these go in opposite directions, they cancelled out in 2.5, which is why attr_gap looked like it didn’t matter.

The partner ratings stayed about the same, and gender and age are still not significant. That’s expected, since we only changed how the gap variables work and didn’t touch the partner ratings.

The extended model fits better. AIC dropped from 6418.4 to 6408.5, and the likelihood ratio test gives p = 0.0013. But the AUC barely moved, so the new model doesn’t really predict likes any better.

d)
The extension showed us that overconfidence doesn’t work the same way for men and women. In 2.5 it looked like overrating your intelligence hurt everyone, but it turns out this is mostly true for men. For women, the intelligence gap doesn’t really matter. This is what we expected, since men are being judged by women, and in Task 1 women cared more about intelligence.

The attractiveness gap also works differently for men and women, which we couldn’t see in 2.5. The effect is lower for women than for men, which partly fits what we expected. But the interaction only tells us the two effects are different from each other, not that each one is significant on its own, so we don’t want to read too much into it.

Overall, this changes our conclusion a bit. Instead of saying overconfidence about intelligence hurts everyone, we’d say overconfidence matters differently by gender, and the clearest case is men overrating their intelligence. We still don’t see overconfidence mattering across all four traits.

e)
Every date in this dataset is between a man and a woman, so the participant’s gender and the partner’s gender always go together. So when overrating intelligence hurts men more, we can’t tell if it’s about the men themselves or about the women judging them. We assumed it’s because of what the partner cares about, but the model can’t actually separate the two.

To test this better, we could use the partner’s own stated preferences. The dataset has a variable for how much the partner says they care about intelligence, and we could interact it with intel_gap to see if overrating your intelligence hurts more when your partner cares a lot about it. Data with same sex dates would also help separate the two.
